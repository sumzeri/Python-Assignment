```python
#6
#Aim:To write a Python program to check whether a given year is a leap year.
year = int(input("Enter a year: "))
if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
#14
#To write a Python program to reverse the digits of a given number.
n = int(input("Enter a number: "))
r = 0
while n > 0:
    d = n % 10
    r = r * 10 + d
    n //= 10
print("Reversed number =", r)
```

```python
#15
#Aim:To write a Python program to count the number of digits in a given integer.
n = int(input("Enter a number: "))
c = 0
while n > 0:
    n //= 10
    c += 1
print("Number of digits =", c)
```

```python
#16
#Aim:To write a Python program to calculate the sum of digits of a given number.
n = int(input("Enter a number: "))
s = 0
while n > 0:
    d = n % 10
    s += d
    n //= 10
print("Sum of digits =", s)
```

```python
#17
#Aim:To write a Python program to check whether a given number is prime.
num = int(input("Enter a number: "))
if num > 1:
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
else:
    print("Not Prime")
```

```python
#18
#Aim:To write a Python program to generate the Fibonacci series up to N terms.
n = int(input("Enter the number of terms: "))
a, b = 0, 1
print("Fibonacci series:")
for i in range(n):
    print(a)
    a, b = b, a + b
```

```python
#19
#Aim:To write a Python program to check whether a string is a palindrome.
s = input("Enter a string: ")
if s == s[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
#20
#Aim:To write a Python program to check whether a given number is an Armstrong number.
n = int(input("Enter a number: "))
t = n
ds = len(str(n))
s = 0
while t > 0:
    d = t % 10
    s += d ** ds
    t //= 10
if s == n:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```
