```python
#Q5
numbers = [10, 21, 4, 45, 66, 93, 11]
even = 0
odd = 0
for n in numbers:
    if n % 2 == 0:
        even += 1
    else:
        odd += 1
print("Even:", even)
print("Odd:", odd)
```

```python
#Q12
car = {
    "brand": "Toyota",
    "model": "Camry",
    "year": 2022,
    "color": "blue"
}
del car["color"]
print("Key-value pairs:", car.items())
if "brand" in car:
    print("Brand key exists")
else:
    print("Brand key does not exist")
```
